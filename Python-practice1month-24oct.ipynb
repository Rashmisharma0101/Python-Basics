```python
df = pd.read_csv(r'C:\Users\rasha\OneDrive\Desktop\Job 2026\ML PROJECTS\AirBNB Price prediction\AB_US_2023.csv')
```

```python
# Recheck missing values, duplicates, and inconsistent data (like ‘NY’ vs ‘New York’). 
# Handle with df.drop_duplicates(), fillna(), and string cleanup (str.strip(), str.lower()).
df = df[:10000]
print(df.isnull().sum())
df = df.drop_duplicates()

```

```python
df = df.fillna({'reviews_per_month': df['reviews_per_month'].mean(), 'last_review':  0})
```

```python
df['neighbourhood'].value_counts().sort_index()
```

```python
# Presidio   and Presidio Heights can be same
df['neighbourhood'] = df['neighbourhood'].replace('Presidio Heights', 'Presidio')
```

```python
cat = ['neighbourhood_group', 'neighbourhood', 'room_type']
for i in cat:
    df[i] = df[i].str.strip()
    df[i] = df[i].str.lower()
```

```python
# Filter rows using conditions (e.g., df[df['price'] > 500]), query(), and boolean indexing.

df_filtered = df[df['price']> 500]
print(df_filtered.shape)

df_query = df.query('price > 500')
print(df_query.shape)
```

```python
# Sort values (sort_values()) and rank items (rank()) to identify top-performing categories or customers.

df_rank_frame = pd.DataFrame(df.groupby('room_type')['price'].sum().sort_values(ascending = False).reset_index())
```

```python
print(df_rank_frame)
df_rank_frame['rank'] = df_rank_frame['price'].rank(method = 'dense', ascending = False)
df_rank_frame
```

```python
# Use groupby() to find average, median, and count metrics (e.g., df.groupby('neighbourhood')['price'].mean()). 
# Combine multiple aggregations using .agg().

df_agg = (df.groupby('neighbourhood').agg(price_mean = ('price', 'mean'),
                                price_max = ('price', 'max'), 
                                total_reviews = ('reviews_per_month', 'sum')).reset_index())
```

```python
df_agg
```

```python
# Create pivot tables using pd.pivot_table() for multidimensional summaries (like Excel-style analysis).
pd.pivot_table(
    df,
    index='neighbourhood',
    columns='room_type',
    values='price',
    aggfunc=['mean', 'max', 'min']
)
```

```python
# Practice merging two DataFrames using merge(), join(), and concat() on different keys.

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'city': ['Delhi', 'Mumbai', 'Chennai', 'Pune']
})

# Second DataFrame — Orders
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_id': [1, 2, 1, 3, 2],
    'amount': [250, 400, 150, 600, 350]
})

# join dataframes
merged = customers.merge(orders, how = 'left', on = 'customer_id')
merged
```

```python
# join needs column to be index, does not work on normal column

joined_df = customers.set_index('customer_id').join(
    orders.set_index('customer_id'), how = 'left').reset_index()

joined_df
```

```python
# merge and join are like sql operations, concat is smply appending dataframes along the specified axis
import pandas as pd

df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'city': ['Delhi', 'Mumbai', 'Chennai']
})

df2 = pd.DataFrame({
    'id': [4, 5],
    'city': ['Pune', 'Kolkata']
})

combined = pd.concat([df1, df2], axis=0)
print(combined)
```
